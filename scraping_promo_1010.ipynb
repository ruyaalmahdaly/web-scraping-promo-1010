```python

import pandas as pd
from pathlib import Path

file_bersih = Path("/content/output/dataset_promo_1010_indonesia.csv")
df = pd.read_csv(file_bersih)

# Kata kunci untuk pemeriksaan relevansi awal
kata_kunci = [
    "10.10", "10/10", "tanggal kembar",
    "promo", "diskon", "voucher", "gratis ongkir",
    "lazada", "shopee", "tokopedia", "blibli",
    "belanja", "flash sale", "cashback"
]

def cek_relevansi(teks):
    teks = str(teks).lower()
    return any(kata in teks for kata in kata_kunci)

df["indikasi_relevan"] = df["full_text"].apply(cek_relevansi)

print("Total cuitan:", len(df))
print("Terindikasi relevan:", int(df["indikasi_relevan"].sum()))
print("Perlu diperiksa manual:", int((~df["indikasi_relevan"]).sum()))

# Tampilkan seluruh cuitan agar bisa diperiksa
display(df[["created_at", "full_text", "indikasi_relevan"]])
```

```python

import pandas as pd
from pathlib import Path

folder = Path("/content/output")
file_asal = folder / "dataset_promo_1010_indonesia.csv"

df = pd.read_csv(file_asal)
df["full_text"] = df["full_text"].fillna("").astype(str)
teks = df["full_text"].str.lower()

# Indikasi hubungan langsung dengan promo tanggal kembar
pola_1010 = r"10[./ -]?10|tanggal kembar|10 oktober"
pola_promo = r"promo|diskon|voucher|gratis ongkir|cashback|flash sale|belanja|lazada|shopee|tokopedia|traveloka|pahamify"

mask = teks.str.contains(pola_1010, regex=True) & teks.str.contains(pola_promo, regex=True)
df_relevan = df[mask].copy()

jumlah_sebelum = len(df_relevan)

# Hapus duplikat berdasarkan ID, kemudian teks
if "id_str" in df_relevan.columns:
    df_relevan = df_relevan.drop_duplicates(subset=["id_str"])

df_relevan = df_relevan.drop_duplicates(subset=["full_text"])

file_final = folder / "dataset_promo_1010_relevan.csv"
df_relevan.to_csv(file_final, index=False, encoding="utf-8-sig")

print("Data awal:", len(df))
print("Terpilih sebelum hapus duplikat:", jumlah_sebelum)
print("Data relevan unik:", len(df_relevan))
print("Lokasi file:", file_final)

display(df_relevan[["created_at", "full_text"]])
```

```python

import re
import pandas as pd

def bersihkan_teks(teks):
    teks = str(teks).lower()
    teks = re.sub(r'https?://\S+|www\.\S+', ' ', teks)
    teks = re.sub(r'@\w+', ' ', teks)
    teks = re.sub(r'&amp;', 'dan', teks)
    teks = re.sub(r'[^a-z0-9\s]', ' ', teks)
    teks = re.sub(r'\s+', ' ', teks).strip()
    return teks

df["teks_normalisasi"] = df["full_text"].apply(bersihkan_teks)

print("Jumlah data awal:", len(df))
print("Jumlah teks unik:", df["teks_normalisasi"].nunique())
print("Jumlah teks berulang:",
      len(df) - df["teks_normalisasi"].nunique())

print("\nContoh teks yang berulang:")
frekuensi = df["teks_normalisasi"].value_counts()
display(frekuensi[frekuensi > 1].reset_index().rename(
    columns={
        "index": "teks_normalisasi",
        "teks_normalisasi": "jumlah"
    }
))
```
